E4.1 A cantilever beam has a length L = 50 mm, a width b = 5 mm, and a
thickness t = 1 mm. It is made of an aluminum alloy. By how much will
the end deflect under an end-load of 5 N? Use data from Appendix A.4 for the
(mean) value of Young’s modulus of aluminum alloys, the equation for the
elastic deflection of a cantilever from Appendix B.3, and for the second
moment of a beam from Appendix B.2 to find out.

```python
#datos
L = 50e-3
b = 5e-3
t = 1e-3
F = 5
#Apendice A.4
E_Al = ((68+82)/2)*1e9
#Apendice B.2
I = b*t**3/12
c1 = 3
#Preguntas
# theta =?
```

Preguntas
* $\theta$ = $\frac{FL^2}{C_1EI}$
